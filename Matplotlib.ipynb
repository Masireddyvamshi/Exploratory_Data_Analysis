# LinePlot

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [5, 6, 7, 8]
plt.figure(facecolor='lightblue')
plt.plot(
    x, y,
    color='red',          # must be lowercase
    linestyle='--',
    linewidth=3,          # should be int/float, not string
    marker='s',           # use lowercase 'o'
    markersize=5,         # should be int/float
    markerfacecolor='blue',
    markeredgecolor='black',
    label='Example',
    alpha=0.8           # must be between 0 and 1
)
plt.grid()
plt.legend()
plt.show()
```

```python
months=[1,2,3,4,5]
product_A=[20,30,40,50,60]
product_B=[100,200,300,400,500] 
plt.plot(months,product_A,label='Product A',color='Black',marker='x')
plt.plot(months,product_B,label='Product B',color='Orange',marker='x')
plt.xticks(months)
plt.yticks([i for i in range(100,1100,100)])
plt.legend()
plt.show()
```

```python
# Both products show an upward trend over months
# product b consistently records higher saler than product a
# product a shows slow but steady growthh
# product b shows rapid and strong growth
# No flutuations or declines are observed in the sales of both products
```

```python
x=[1,2,3,4,5]
A=[10,50,10,50,10]
B=[10,20,30,40,50]
plt.plot(x,A,label='Product A',color='red',marker='*')
plt.plot(x,B,label='Product B',color='blue',marker='*')
plt.legend()
plt.show()
```

```python
# The product b sales are consistent and steady growth
# The product a sales are increasing rapidly and decreasing rapidly over the period of time
# There is fluctuations in product a there is no consitency growth in sales compared to b
```

```python
x = [1, 2, 3, 4, 5]
A = [100, 90, 80, 70, 60]
B = [50, 70, 90, 110, 130]

plt.plot(x, A, label="Product A")
plt.plot(x, B, label="Product B") 
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
A = [1, 2, 3, 4, 5]
B = [100, 200, 300, 400, 500]

plt.plot(x, A, label="Product A")
plt.plot(x, B, label="Product B")
plt.legend()
plt.show()
```

```python
# Both products show proportional linear growth
# The growth pattern is same, but magnitude is different
# Product B has higher volume, not necessarily a higher growth rate
```

# Histogram

```python
data = [45, 50, 55, 60, 62, 65, 70, 75, 80]

plt.hist(
    data,
    bins=5,
    color='skyblue',
    edgecolor='black',
    alpha=0.8
)
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram Example') 
plt.show()  
```

```python
df=pd.read_csv(r"C:\Users\masir\Downloads\Datasets\Online Sales Data.csv")
```

```python
df.head()
```

```python
plt.hist(
    df['Units Sold'],
    bins=5,
    range=(1, 10),
    density=False,
    color='Red',
    edgecolor='green',
    linewidth=4,
    alpha=1,
    label='Units_sold',
    histtype='bar',
    orientation='vertical'
)

plt.legend()
plt.grid()
plt.show()
```

```python
# Most of the Units Sold values are concentrated between 1 and 3.
# Units sold less than 2 are close to 100, indicating high frequency at low values.
# The distribution is right-skewed (positively skewed), with a long tail towards higher values.
# Very few observations occur above 6, showing possible outliers or rare high sales.
```

```python
plt.hist(
    df['Units Sold'],
    bins=10,
    range=(1, 10),
    density=True,
    color='Red',
    edgecolor='green',
    linewidth=4,
    alpha=1,
    label='Units_sold',
    histtype='step',
    orientation='vertical'
)

plt.legend()
plt.grid()
plt.show()
```

```python
plt.hist(
    df['Units Sold'],
    bins=10,
    range=(1, 10),
    density=False,
    color='Red',
    edgecolor='green',
    linewidth=4,
    alpha=1,
    label='Units_sold',
    histtype='stepfilled',
    orientation='vertical'
)

plt.legend()
plt.grid()
plt.show()
```

# Line Plot

```python


age = [22, 25, 28, 32, 36, 40, 45]
salary = [25000, 38000, 36000, 55000, 52000, 80000, 78000]  # fluctuations

plt.plot(
    age,
    salary,
    marker='o',
    linestyle='-',
    linewidth=2,
    alpha=0.8,
    label='Salary Trend'
)

plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary with Fluctuations')
plt.legend()
plt.show()
```

```python
# 📈 Overall upward trend → career growth

# 🔻 Temporary drops → job switch / gap

# 🔺 Sharp rise → promotion or role change
```
