# Series
##### Series	Homogeneous in storage (one dtype)
##### Iteration	Shows individual element datatypes
##### object dtype	A general “box” that can hold mixed Python objects

```python
arr=np.random.randint(1,50,5)
print(arr)
```

```python
series=pd.Series(arr)
print(series)
print(type(series))
```

```python
series=pd.Series(arr,name='integers',index=[chr(65+i) for i in range(5)])
print(series)
print(type(series))
```

```python
# Tabular Dataset -------> 2D array
# Images ----------------> 2D array or 3D array
# Text Dataset ----------> 1D array
# Audio Dataset ---------> 1D array
# Sensor Data -----------> 1D array
```

# dtypes and dtype

```python
marks = pd.Series(
    data={'Math': 90, 'Science': 85, 'English': 70, 'History': 60},
    index=['Math', 'Science', 'English', 'History'],
    dtype='int64',
    name='Exam Marks'
)
print(marks.dtype)
```

# indexing and slicing

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'], name='Numbers')
print(s)
```

```python
print(s['a'])
print(s['b'])
print(s['c'])
print(s['d'])
print(s['e'])
```

```python
print(s[['a','c','e']])
```

```python
print(s[0])     
print(s[[0, 2, 4]])  
```

```python
print(s[0:3])
```

```python
print(s['a':'c'])
```

```python
print(s[0::2])
```

```python
print(s['a'::2]) 
```

```python
print(s[s>25])
```
