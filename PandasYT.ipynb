# astype()

```python
# Step	What happens	Effect
# arr1 = pd.Series(np.random.randint(10,30,10000))	Creates 10,000 random integers in memory (default dtype = int64)	Stored in binary
# arr2 = arr1.astype(np.int8)	Converts to 8-bit integers (saves memory in RAM)	Takes 1/8th of space in memory, not on disk
# .to_csv()	Converts all numbers to text (strings) before saving	Every number like 27 becomes '27' (2 characters)
# Disk file (.csv)	Contains plain text digits and commas/newlines	File size same regardless of dtype


 # let us create a series array with 10000 datapoints between 10,30 and then save it 
# let us convert the same array into int8 and then save it 
# and compare the memory occupied by both of the files

# what is the benifit of type casting
# memory consumption is optimized
# speed ----> execution | time taken to load the data is also less

# “Pickle is 1D array where CSV is 2D”
```

```python
arr1 = pd.Series(np.random.randint(10,30,10000),name="age") 
arr1.to_csv(path+'/' + 'age_seriesint.csv') 
```

```python
arr2=arr1.astype(np.int8)
arr2.to_csv(path+'/' + 'age_seriesint8.csv')
```

```python
arr1 = pd.Series(np.random.randint(10,30,10000),name="age") 
print(arr1.nbytes)
print(arr1.dtype)
arr1.to_pickle(path+'/' + 'age_seriesint.pkl')
```

```python
arr2=arr1.astype(np.int8)
print(arr2.nbytes)
print(arr2.dtype)
arr2.to_pickle(path+'/' + 'age_seriesint8.pkl')
```
