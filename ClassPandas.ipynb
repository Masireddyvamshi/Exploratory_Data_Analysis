```python
d={'names':[],'genders':[],'ages':[],'countrys':[],'states':[],'lattitudes':[],'longitudes':[]}
while len(d['names'])<=1000:
    data=requests.get('https://randomuser.me/api/')
    if data.status_code == 200:
        try:
            json_data=data.json()
            named=json_data['results'][0]['name'] 
            name=f'{named['title']} {named['first']} {named['last']}'
            d['names'].append(name)
            gender=json_data['results'][0]['gender']
            d['genders'].append(gender)
            age=json_data['results'][0]['dob']['age']
            d['ages'].append(age)
            country=json_data['results'][0]['location']['country']
            d['countrys'].append(country)
            state=json_data['results'][0]['location']['state']
            d['states'].append(state)
            lattitude=json_data['results'][0]['location']['coordinates']['latitude']
            d['lattitudes'].append(lattitude)
            longtitude=json_data['results'][0]['location']['coordinates']['longitude']
            d['longitudes'].append(lattitude)
        except IndexError:
            continue
```
